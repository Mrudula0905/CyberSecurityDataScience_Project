````python
import torch
from torch import nn
````

````python
device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using device:", device)
````

````python
class VulnPredictModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.linear_stack = nn.Sequential(
            nn.Linear(768, 64),
            nn.ReLU(),

            nn.Linear(64, 64),
            nn.ReLU(),

            nn.Linear(64, 1),
            nn.Sigmoid()
        )

    def forward(self, x):

        pred = self.linear_stack(x)

        return pred
````

````python
model = VulnPredictModel().to(device)
````

````python
model.load_state_dict(
    torch.load(
        "model_2023-03-28_20-03.pth",
        map_location=device
    )
)
````

````python
model.eval()

print("Model loaded successfully!")
````

###*Task 5*

Make a prediction on the provided dataset and compute the following values:
- True Positives
- True Negatives
- False Positives
- False Negatives

````python
# TODO: makethe prediction for all the samples in the validation set.

X = X.squeeze(1)

X_tensor = torch.tensor(X, dtype=torch.float32).to(device)

y_tensor = torch.tensor(y, dtype=torch.float32).to(device)

with torch.no_grad():

    outputs = model(X_tensor)


preds = (outputs > 0.5).float()


preds = preds.squeeze()

y_tensor = y_tensor.squeeze()


tp = ((preds == 1) & (y_tensor == 1)).sum().item()

tn = ((preds == 0) & (y_tensor == 0)).sum().item()

fp = ((preds == 1) & (y_tensor == 0)).sum().item()

fn = ((preds == 0) & (y_tensor == 1)).sum().item()


print("True Positives :", tp)

print("True Negatives :", tn)

print("False Positives:", fp)

print("False Negatives:", fn)
# todo: compute true positives, true negatives, false postives and false negatives.
````

### *Task 6*

Compute the corresponding performance metrics **manually** (do not use PyTorch's predefined metrics):
- Accuracy
- Precision
- Recall
- F1

````python

accuracy = (tp + tn) / (tp + tn + fp + fn)


precision = tp / (tp + fp)


recall = tp / (tp + fn)


f1 = 2 * (precision * recall) / (precision + recall)


print("Accuracy :", accuracy)

print("Precision:", precision)

print("Recall   :", recall)

print("F1 Score :", f1)
````
